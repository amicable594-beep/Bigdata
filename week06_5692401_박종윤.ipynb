## 과제 내용
은닉층이 1개인 모델과 2개인 모델을 각각 작성하고 구조 차이를 설명합니다.

## 제출 방법
모든 셀을 실행해 출력이 보이는 상태로 저장한 뒤 `week06_학번_이름.ipynb` 로 제출합니다.


```python
import torch
import torch.nn as nn
import pandas as pd

torch.manual_seed(0)
```

## 문제 1. 은닉층 1개 모델 (25점)
`nn.Module` 을 상속해 아래 조건의 모델을 작성하세요.

- 입력 특성 5개, 은닉층 뉴런 16개, 출력 3개
- `__init__` 과 `forward` 를 모두 구현하고 각 줄에 주석
- `print(model)` 로 구조 출력


```python
# 답안
# nn.Module을 상속받아 모델 클래스 정의
class Model1(nn.Module):

    # 모델의 층을 초기화하는 함수
    def __init__(self):
        # 부모 클래스 초기화
        super(Model1, self).__init__()

        # 입력 5개, 은닉층 뉴런 16개
        self.fc1 = nn.Linear(5, 16)

        # 은닉층 뉴런 16개, 출력 3개
        self.fc2 = nn.Linear(16, 3)

        # ReLU 활성화 함수 정의
        self.relu = nn.ReLU()

    # 순전파 함수 정의
    def forward(self, x):
        # 첫 번째 은닉층 통과
        x = self.fc1(x)

        # ReLU 활성화 함수 적용
        x = self.relu(x)

        # 출력층 통과
        x = self.fc2(x)

        # 최종 결과 반환
        return x

# 모델 생성
model1 = Model1()

# 모델 구조 출력
print(model1)
```

## 문제 2. 은닉층 2개 모델 (25점)
문제 1과 같은 입력·출력에 은닉층을 2개로 늘린 모델을 작성하고 구조를 출력하세요.


```python
# 답안
# nn.Module을 상속받아 모델 클래스 정의
class Model2(nn.Module):

    # 모델의 층을 초기화하는 함수
    def __init__(self):
        # 부모 클래스 초기화
        super(Model2, self).__init__()

        # 입력 5개, 첫 번째 은닉층 뉴런 16개
        self.fc1 = nn.Linear(5, 16)

        # 첫 번째 은닉층 16개, 두 번째 은닉층 16개
        self.fc2 = nn.Linear(16, 16)

        # 두 번째 은닉층 16개, 출력 3개
        self.fc3 = nn.Linear(16, 3)

        # ReLU 활성화 함수 정의
        self.relu = nn.ReLU()

    # 순전파 함수 정의
    def forward(self, x):
        # 첫 번째 은닉층 통과
        x = self.fc1(x)

        # ReLU 활성화 함수 적용
        x = self.relu(x)

        # 두 번째 은닉층 통과
        x = self.fc2(x)

        # ReLU 활성화 함수 적용
        x = self.relu(x)

        # 출력층 통과
        x = self.fc3(x)

        # 최종 결과 반환
        return x

# 모델 생성
model2 = Model2()

# 모델 구조 출력
print(model2)
```

## 문제 3. 두 모델의 구조 비교표 (30점)
아래 항목으로 표를 만들어 출력하세요.

| 항목 | 은닉층 1개 | 은닉층 2개 |
| --- | --- | --- |
| Linear 층 수 | | |
| 전체 파라미터 수 | | |
| 활성화 함수 개수 | | |
| 출력 shape | | |


```python
# 답안 — pd.DataFrame 으로 만들어 출력하세요
# 전체 파라미터 수 계산 함수
def count_parameters(model):
    return sum(p.numel() for p in model.parameters())

# Linear 층 개수 계산
linear1 = sum(1 for m in model1.modules() if isinstance(m, nn.Linear))
linear2 = sum(1 for m in model2.modules() if isinstance(m, nn.Linear))

# 활성화 함수 개수 계산
relu1 = sum(1 for m in model1.modules() if isinstance(m, nn.ReLU))
relu2 = sum(1 for m in model2.modules() if isinstance(m, nn.ReLU))

# 임의 입력 생성
x = torch.randn(1, 5)

# 각 모델의 출력 shape 확인
shape1 = model1(x).shape
shape2 = model2(x).shape

# 비교표 출력
print("항목\t\t은닉층 1개\t은닉층 2개")
print("-" * 50)
print(f"Linear 층 수\t{linear1}\t\t{linear2}")
print(f"전체 파라미터 수\t{count_parameters(model1)}\t\t{count_parameters(model2)}")
print(f"활성화 함수 개수\t{relu1}\t\t{relu2}")
print(f"출력 shape\t{shape1}\t{shape2}")
```

## 문제 4. 구조 차이 설명 (20점)
아래 세 가지를 각각 두세 줄로 설명하세요.

1. 층을 늘리면 파라미터 수가 왜 늘어나는가
2. 활성화 함수가 없으면 층을 쌓아도 소용없는 이유
3. 층을 늘리는 것이 항상 좋은가 (근거를 함께)


**답:**

1.층을 추가하면 새로운 가중치와 편향이 필요하기 때문에 파라미터 수가 증가한다. 각 층의 입력과 출력 뉴런 수에 따라 학습해야 하는 파라미터가 늘어난다.

2.활성화 함수가 없으면 여러 층을 쌓아도 결국 하나의 선형 변환과 같다. ReLU와 같은 활성화 함수를 사용해야 비선형적인 복잡한 패턴을 학습할 수 있다.

3.층을 늘리면 복잡한 데이터의 특징을 학습할 수 있지만 항상 좋은 것은 아니다. 파라미터가 증가하면 학습 시간이 길어지고 과적합이 발생할 수 있으므로 데이터에 적합한 층의 수를 선택해야 한다.


## 문제 5. 임의 입력으로 확인 (보너스)
두 모델에 각각 `torch.randn(8, 5)` 를 넣어 출력 shape을 확인하고, 중간 층의 shape 변화를 출력하세요.


```python
# 답안
# 임의의 입력 데이터 생성 (8개 데이터, 특성 5개)
x = torch.randn(8, 5)

# 문제 1 모델 확인
print("===== 은닉층 1개 모델 =====")

# 첫 번째 은닉층 통과
h1 = model1.fc1(x)
print("은닉층 shape:", h1.shape)

# 활성화 함수 적용
h1 = model1.relu(h1)

# 출력층 통과
out1 = model1.fc2(h1)
print("최종 출력 shape:", out1.shape)


# 문제 2 모델 확인
print("\n===== 은닉층 2개 모델 =====")

# 첫 번째 은닉층 통과
h2 = model2.fc1(x)
print("첫 번째 은닉층 shape:", h2.shape)

# 활성화 함수 적용
h2 = model2.relu(h2)

# 두 번째 은닉층 통과
h3 = model2.fc2(h2)
print("두 번째 은닉층 shape:", h3.shape)

# 활성화 함수 적용
h3 = model2.relu(h3)

# 출력층 통과
out2 = model2.fc3(h3)
print("최종 출력 shape:", out2.shape)
```
