## 문제 1. Tensor 5개 생성과 속성 확인표 (40점)
서로 다른 Tensor 5개를 만들고, 아래 항목을 담은 표를 출력하세요.

| 이름 | shape | dim | dtype | device |

각 Tensor를 왜 그렇게 만들었는지 주석으로 한 줄 적으세요.


```python
# 답안

# 1. 숫자 하나를 저장하는 0차원 Tensor를 만들기 위해 생성
t1 = torch.tensor(7)

# 2. 숫자 세 개를 일렬로 저장하는 1차원 Tensor를 만들기 위해 생성
t2 = torch.tensor([1, 2, 3])

# 3. 행과 열이 있는 2차원 Tensor를 만들기 위해 생성
t3 = torch.tensor([[1, 2], [3, 4]])

# 4. 0으로 채워진 2행 3열 Tensor를 만들기 위해 생성
t4 = torch.zeros((2, 3))

# 5. 1부터 6까지의 숫자를 2행 3열로 배열한 Tensor를 만들기 위해 생성
t5 = torch.arange(1, 7).reshape(2, 3)


# Tensor 5개를 이름과 함께 저장
tensors = {
    "t1": t1,
    "t2": t2,
    "t3": t3,
    "t4": t4,
    "t5": t5
}

# 각 Tensor의 정보를 모을 빈 리스트
result = []

# Tensor를 하나씩 꺼내 속성을 기록
for name, tensor in tensors.items():
    result.append({
        "이름": name,
        "shape": tuple(tensor.shape),
        "dim": tensor.dim(),
        "dtype": tensor.dtype,
        "device": tensor.device
    })

# 결과를 표로 변환하여 출력
df = pd.DataFrame(result)
print(df.to_string(index=False))
```

## 문제 2. Tensor 연산 (20점)
(3, 4) 와 (4, 2) Tensor를 만들어 아래를 출력하세요.
- 행렬곱 결과와 그 shape
- 첫 번째 Tensor의 열별 평균과 행별 합계


```python
# 답안


import torch

# 1. 1부터 12까지의 숫자로 3행 4열 Tensor 생성
A = torch.arange(1, 13).reshape(3, 4)

# 2. 1부터 8까지의 숫자로 4행 2열 Tensor 생성
B = torch.arange(1, 9).reshape(4, 2)

# 3. A와 B의 행렬곱 계산
C = torch.matmul(A, B)

# 4. A의 열별 평균 계산
column_mean = A.mean(dim=0)

# 5. A의 행별 합계 계산
row_sum = A.sum(dim=1)

# 결과 출력
print("Tensor A:")
print(A)

print("\nTensor B:")
print(B)

print("\n행렬곱 A × B:")
print(C)

print("\nA의 열별 평균:")
print(column_mean)

print("\nA의 행별 합계:")
print(row_sum)
```

## 문제 3. CPU와 GPU 비교 (20점)
같은 크기의 행렬곱을 CPU와 GPU에서 각각 실행해 걸린 시간을 출력하고, 차이를 한두 줄로 설명하세요.
GPU를 쓸 수 없는 경우에는 CPU 결과와 그 이유를 적으세요.


```python
# 답안
import torch

import torch
import time

# 1. CPU에서 사용할 Tensor 생성
A_cpu = torch.randn(2000, 2000)
B_cpu = torch.randn(2000, 2000)

# 2. CPU 행렬곱 및 실행 시간 측정
start = time.time()
C_cpu = torch.matmul(A_cpu, B_cpu)
cpu_time = time.time() - start

# 3. GPU 사용 가능 여부 확인
if torch.cuda.is_available():

    # Tensor를 GPU로 이동
    A_gpu = A_cpu.to("cuda")
    B_gpu = B_cpu.to("cuda")

    # GPU 연산 시간 측정
    torch.cuda.synchronize()
    start = time.time()

    C_gpu = torch.matmul(A_gpu, B_gpu)

    torch.cuda.synchronize()
    gpu_time = time.time() - start

    # 4. 두 실행 시간 출력
    print("===== 실행 결과 =====")
    print("CPU 실행 시간:", round(cpu_time, 4), "초")
    print("GPU 실행 시간:", round(gpu_time, 4), "초")

    # 5. 속도 비교
    if gpu_time > 0:
        print("GPU 속도 배율:", round(cpu_time / gpu_time, 2), "배")

    if cpu_time > gpu_time:
        print("결과: GPU가 더 빠릅니다.")
    elif cpu_time < gpu_time:
        print("결과: CPU가 더 빠릅니다.")
    else:
        print("결과: 실행 시간이 같습니다.")

else:
    print("GPU를 사용할 수 없습니다.")
    print("CPU 실행 시간:", round(cpu_time, 4), "초")
```

## 문제 4. 자동미분 (20점)
아래 두 가지를 하세요.
1. $y = 2x^3 + 5x$ 에서 $x=3$ 일 때의 도함수를 자동미분으로 구하고, 손으로 구한 값과 비교하세요.
2. 자기가 만든 수식 하나에 대해 미분값을 구하고, 손 계산과 일치하는지 확인하세요.


```python
# 답안


import torch

# 1. y = 2x^3 + 5x, x = 3
x = torch.tensor(3.0, requires_grad=True)
y = 2 * x**3 + 5 * x

y.backward()

print("1번")
print("자동미분:", x.grad.item())

hand1 = 6 * 3**2 + 5
print("손 계산:", hand1)
print("결과:", x.grad.item() == hand1)


# 2. y = 4x^3 - 2x^2 + 3x + 1, x = 2
x = torch.tensor(2.0, requires_grad=True)
y = 4 * x**3 - 2 * x**2 + 3 * x + 1

y.backward()

print("\n2번")
print("자동미분:", x.grad.item())

hand2 = 12 * 2**2 - 4 * 2 + 3
print("손 계산:", hand2)
print("결과:", x.grad.item() == hand2)
```
